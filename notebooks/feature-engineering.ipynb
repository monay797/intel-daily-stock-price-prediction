```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"C:\Users\Monay\Code\Machine Learning\Projects\intel-daily-stock-price-prediction\data\cleaned\intel-daily-stock-price-data-cleaned.csv")
```

```python
df.head()
```

```python
df.tail()
```

```python
df["MA5"] = df["Close"].rolling(window=5).mean()
df.head()
```

```python
df["MA14"] = df["Close"].rolling(window=14).mean()
df.head()
```

```python
df["High-Low Range"] = df["High"] - df["Low"]
df.head()
```

```python
df["Target"] = df["Close"].shift(-1)
df.head()
```

```python
df.tail()
```

```python
df = df.drop(columns=["Volume", "Dividends"])
df.head()
```

```python
df = df.dropna()
df
```

```python
df.set_index('Date', inplace=True)
df
```

```python
import matplotlib.pyplot as plt

corr = df.corr(numeric_only=True)

plt.figure(figsize=(8,6))

plt.imshow(corr, cmap='coolwarm')

plt.colorbar()

plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title("Correlation Matrix")

plt.show()
```
