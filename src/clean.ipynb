```python
import pandas as pd

# Load Data
df = pd.read_csv(r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\raw\intel-daily-stock-price-data-raw.csv")

# Dropping Unnecessary Rows
df = df.drop([0, 1])

# Changing Column Name
df = df.rename(columns={"Price":"Date"})

# Date Conversion
df["Date"] = pd.to_datetime(df["Date"])

# Converting Columns Into Numerical Values (Float/Int)
numeric_cols = ["Open", "High", "Low", "Close", "Volume", "Dividends"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col])

# Sorting Date
df = df.sort_values("Date")

# Row Index Reset
df = df.reset_index(drop=True)

# Data Frame Information
df.info()
```

```python
df.to_csv(
    r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\cleaned\intel-daily-stock-price-data-cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
