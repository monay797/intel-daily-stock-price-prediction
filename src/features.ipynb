```python
import pandas as pd

# Load Data
df = pd.read_csv(r"C:\Users\Monay\Code\Machine Learning\Projects\intel-daily-stock-price-prediction\data\cleaned\intel-daily-stock-price-data-cleaned.csv")

# 5-Days Moving Average Feature
df["MA5"] = df["Close"].rolling(window=5).mean()

# 14-Days Moving Average Feature
df["MA14"] = df["Close"].rolling(window=14).mean()

# High-Low Range Feature
df["High-Low Range"] = df["High"] - df["Low"]

# Target Feature
df["Target"] = df["Close"].shift(-1)

# Dropping Unnecessary Columns/Features
df = df.drop(columns=["Volume", "Dividends"])

# Dropping Rows with N/A
df = df.dropna()

# Switching the Date Column as Index
df.set_index('Date', inplace=True)
```

```python
df.to_csv(r"C:\Users\Monay\Code\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\intel-daily-stock-price-data-processed.csv")

print("Processed dataset saved successfully.")
```
