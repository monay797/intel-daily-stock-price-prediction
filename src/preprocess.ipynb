```python
import pandas as pd
import joblib
from sklearn.preprocessing import StandardScaler

# Load Data
df = pd.read_csv(
    r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\intel-daily-stock-price-data-processed.csv",
    index_col="Date",
    parse_dates=True
                )

# Initialize X and Y features
X = df.drop("Target", axis=1)
y = df["Target"]

# Splitting the Data
split_data = int(len(X) * 0.8)

# Initialize Train and Test Set
X_train, y_train = X.iloc[:split_data], y.iloc[:split_data]
X_test, y_test = X.iloc[split_data:], y.iloc[split_data:]

# Initialize the Standard Scaler
scaler = StandardScaler()

# Initialize the scaled train and test set
scaler.fit(X_train)

# Saving initialized variables to joblib file
joblib.dump(scaler, r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\models\scaler.joblib")
```

```python
# Create a CSV file for splitted dataset
X_train.to_csv(r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\X_train.csv")
X_test.to_csv(r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\X_test.csv")
y_train.to_csv(r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\y_train.csv")
y_test.to_csv(r"C:\Users\norie\Documents\Machine Learning\Projects\intel-daily-stock-price-prediction\data\processed\y_test.csv")

print("Splitted datasets saved successfully.")
```
